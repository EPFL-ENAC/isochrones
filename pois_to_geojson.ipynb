# Convert SIAE and social centers POIs to GeoJSON

This notebook converts the CSV file containing the geocoded addresses of SIAE and social centers into a GeoJSON file for further use in isopois.

```python
import os
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
```

```python
data_dir = "path/to/data/directory"  # Data is located in 0092-Isochrones
filename = "adresses CDS-CCAS-SIAE.geocoded.csv"
```

```python
# Convert the CSV to GeoJSON
csv_path = os.path.join(data_dir, filename)
df = pd.read_csv(csv_path, encoding="UTF-8", sep=";")

# Filter out unnamed columns that appear due to extra separators
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]
```

```python
geometry = [Point(xy) for xy in zip(df["longitude"], df["latitude"])]
gdf = gpd.GeoDataFrame(df, geometry=geometry)
gdf.crs = (
    "EPSG:4326"  # Temporary CRS for now, waiting for confirmation on the correct one
)
```

```python
gdf.to_file(os.path.join(data_dir, "pois_siae_social.geojson"), driver="GeoJSON")
```
